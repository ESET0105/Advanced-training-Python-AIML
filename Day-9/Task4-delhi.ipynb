## Building ML model to predict next 5 months avg temperature
using Time Series Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA

# Create the time series data from your table
data = {
    2013: [12.07, 16.87, 22.81, 28.90, 33.78, 32.48, 30.66, 29.55, 29.84, 26.13, 18.81, 15.18],
    2014: [13.43, 15.75, 21.59, 28.06, 31.41, 34.77, 32.03, 31.41, 29.76, 26.50, 20.10, 14.82],
    2015: [12.71, 18.79, 21.55, 28.00, 33.35, 32.74, 30.41, 30.33, 30.65, 26.87, 20.75, 14.99],
    2016: [15.14, 19.03, 25.70, 32.55, 34.73, 34.95, 30.88, 31.12, 31.48, 28.96, 22.96, 17.67],
    2017: [10.00]  # Only January 2017 data
}

# Create time series
dates = []
values = []

for year in [2013, 2014, 2015, 2016]:
    for month in range(12):
        dates.append(f"{year}-{month+1}")
        values.append(data[year][month])

# Add January 2017
dates.append("2017-1")
values.append(data[2017][0])

# Create pandas Series
ts = pd.Series(values, index=pd.to_datetime(dates))
ts = ts.dropna()

print("Original Data:")
print(ts)

# Plot the data
plt.figure(figsize=(12, 6))
plt.plot(ts)
plt.title('Monthly Temperature Data')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.show()

# Build ARIMA model (using simple parameters)
# ARIMA(1,1,1) - AR(1), I(1), MA(1)
model = ARIMA(ts, order=(1, 0, 12))
model_fit = model.fit()

print("\nModel Summary:")
print(model_fit.summary())

# Forecast next 5 months
forecast_steps = 5
forecast = model_fit.forecast(steps=forecast_steps)

# Create future dates
last_date = ts.index[-1]
future_dates = pd.date_range(last_date + pd.DateOffset(months=1), 
                            periods=forecast_steps, freq='MS')

print(f"\nNext {forecast_steps} months predictions:")
for date, temp in zip(future_dates, forecast):
    print(f"{date.strftime('%B %Y')}: {temp:.2f}°C")

# Plot results
plt.figure(figsize=(12, 6))
plt.plot(ts, label='Historical Data', linewidth=2)
plt.plot(future_dates, forecast, 'ro-', label='Forecast', linewidth=2)
plt.title('Temperature Forecast using ARIMA(1,1,12)')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.grid(True)
plt.show()
```
