```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import pandas as pd
import numpy as np
```

```python
X,y = make_blobs(n_samples=1000 , centers=3 , n_features=3)
```

```python
plt.scatter(X[:,0],X[:,1] , c=y)
```

```python
## using Kmeans algorithm
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(X,y,test_size=0.2)
```

```python
# fitting kmeans on x,y data
WESS = []
for i in range(1,11):
    kmeans = KMeans(n_clusters=i , init='k-means++')
    kmeans.fit(xtrain)
    WESS.append(kmeans.inertia_)
```

```python
WESS
```

```python
#plot wess (1,11) to no. of clusters
plt.plot(range(1,11),WESS)
plt.xticks(range(1,11))
plt.xlabel("No. of clusters")
plt.ylabel("WESS")
```

```python
kmeans = KMeans(n_clusters=3,init='k-means++')
```

```python
kmeans.fit_predict(xtrain)
```

```python
y_pred = kmeans.predict(xtest)
```

```python
from sklearn.metrics import silhouette_score
```

```python
silhouette_coefficient = []  # if most of the value is positive , then it does lie in that cluster
for k in range(2,11):
  kmeans = KMeans(n_clusters=k , init='k-means++')
  kmeans.fit(xtrain)
  score = silhouette_score(xtrain,kmeans.labels_)
  silhouette_coefficient.append(score)
```

```python
print(silhouette_coefficient)
```

```python
plt.plot(range(2,11),silhouette_coefficient)
plt.xticks(range(2,11))
plt.xlabel("No. of clusters")
plt.ylabel("silhouette coefficient")
```
